```python
#Write a program to find number of occurrence of each letter present in the given string
s=input()
count=dict()
for i in s:
    count[i] = count.get(i, 0) + 1
print(count)
```

```python
#Write a program to find number of occurrence of each vowel present in the given string
s=input()
count=dict()
for i in s.lower():
    if i in ['a','e','i','o','u']:
        count[i]=count.get(i,0)+1
print(count)
```

```python
# Write a python program,which should print tabular column,with header/
#Column as:List,Tuple,Set,Frozenset,Dictionary
#Rows as:Syntax,Ordered,Mutable,Allow Duplicates,Indexwd,Hetrogeneous,Hashable, can be dictionary key, can be nested,Supports slicing ,lookup speed,Typical use

table= {
    "Characteristics": [
        "Syntax",
        "Ordered",
        "Mutable",
        "Allow Duplicates",
        "Indexed",
        "Heterogeneous",
        "Hashable",
        "Can be dict key",
        "Can be nested",
        "Supports slicing",
        "Lookup speed",
        "Stores",
        "Typical use"
    ],
    "List": [
        "[1, 2, 3]",
        "Yes",
        "Yes",
        "Yes",
        "Yes (0, 1...)",
        "Yes",
        "No",
        "No",
        "Yes",
        "Yes (lst[1:3])",
        "O(n) - Slow",
        "Item sequence",
        "Dynamic lists"
    ],
    "Tuple": [
        "(1, 2, 3)",
        "Yes",
        "No",
        "Yes",
        "Yes (0, 1...)",
        "Yes",
        "Yes*",
        "Yes*",
        "Yes",
        "Yes (tup[1:3])",
        "O(n) - Slow",
        "Fixed sequence",
        "Read-only data"
    ],
    "Set": [
        "{1, 2, 3}",
        "No",
        "Yes",
        "No",
        "No",
        "Yes",
        "No",
        "No",
        "Yes",
        "No",
        "O(1) - Fast",
        "Unique items",
        "Deduplication"
    ],
    "Frozenset": [
        "frozenset([1, 2, 3])",
        "No",
        "No",
        "No",
        "No",
        "Yes",
        "Yes",
        "Yes",
        "Yes",
        "No",
        "O(1) - Fast",
        "Fixed unique items",
        "Set keys in dicts"
    ],
    "Dictionary": [
        "{'key': 'value'}",
        "Yes (Python 3.7+)",
        "Yes",
        "Keys: No | Values: Yes",
        "By Key (not position)",
        "Yes",
        "No",
        "No",
        "Yes",
        "No",
        "O(1) - Fast (by key)",
        "Key-Value pairs",
        "Fast lookups / Mappings"
    ]
}

columns=list(table.keys())
n=len(table["Characteristics"])
w= {}
for key in columns:
    max_len = len(key)
    for value in table[key]:
        if len(value) > max_len:
            max_len = len(value)
    w[key] = max_len
for key in columns:
    print("|"+key.center(w[key]+2)+"|",end="")
print("\n")
for i in range(n):
    for key in columns:
        print("|"+table[key][i].center(w[key]+2)+"|",end="")
    print("\n")
print("\n")

print("*Note: Tuples are hashable and can be dictionary keys ONLY IF all elements inside them are also hashable.")
```
